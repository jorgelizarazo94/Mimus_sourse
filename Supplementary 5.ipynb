## Executive Summary

This document synthesises four complementary supplementary materials that together reveal an integrated biological system governing egg incubation in Northern Mockingbirds (*Mimus polyglottos*). The analyses span from nest architecture and egg characteristics through incubation behaviour to the thermal consequences of nest construction, providing a comprehensive mechanistic understanding of how seasonal, behavioural, and architectural factors interact to influence embryonic development.

### Overview of the Four Supplementary Studies

**Supplementary Material 1: Seasonal Variation in Nest Architecture and Egg Traits**
Examines how nest construction materials, nest mass, and egg characteristics vary across the breeding season (March-July). Generalized linear models demonstrate that nesting material composition declines significantly across months (p < 0.05), while egg mass increases (p < 0.001), suggesting an ontogenetic shift in reproductive investment strategies. Sample size: N = 81 nests.

**Supplementary Material 2: Incubation Behavior and Thermal Coordination**
Analyzes how female incubation attendance, recess duration, visit frequency, and thermal coupling evolve across the breeding season. Mixed-effects models reveal strong within-day consistency in incubation patterns (R² = 0.754 for thermal coupling) and near-perfect coordination of visit frequency (R² = 0.999), indicating physiological synchronization between partners. Sample size: N = 81 nests across 17 months of behavioural observations.

**Supplementary Material 3: Circadian and Seasonal Rhythms in Incubation**
Characterizes the temporal structure of incubation using generalized additive mixed models (GAMMs) applied to 149,472 minute-level observations from 519 nest-days. The analysis reveals dominant circadian periodicity (χ² = 9,925.4, p < 2e-16), with incubation probability modulated by time-of-day and season. Nocturnal interruptions (feeding/maintenance bouts) occur in approximately 40% of nests, concentrated in early morning hours. Sample size: 81 nests, 519 complete nest-days.

**Supplementary Material 4: Nest Mass and Thermal Insulation**
Quantifies how nest mass influences egg cooling dynamics during incubation recesses. Exponential cooling models fitted to 157 experimental temperature traces demonstrate that 35g nests provide approximately 20% superior insulation compared to 8g nests, delaying cooling to physiological zero temperature (26°C) by 3.5 minutes (p = 0.003). This thermoregulatory investment has meaningful implications for embryonic development rates.

### Integrated Findings

The four supplementaries collectively demonstrate that Northern Mockingbird reproduction involves coordinated adjustments across multiple biological scales:

1. **Architectural Adaptation (Supp 1 → Supp 4):** Declining nest material investment paradoxically coincides with functional improvements in thermal properties, suggesting adaptive reallocation of resources.

2. **Behavioral Precision (Supp 2 → Supp 3):** The near-perfect coordination of visit frequency observed in Supplementary 2 is mechanistically explained by strong circadian structure in Supplementary 3, permitting synchronized provisioning and incubation duties.

3. **Seasonal Phenology (Supp 1, 2, 3):** Seasonal variation in egg mass, incubation attendance, and recess patterns collectively reflect climate-driven adjustments in breeding strategy, with egg size and nest insulation buffering against environmental thermal variation.

4. **Physiological Coupling (Supp 2, 3, 4):** Behavioral incubation patterns directly interact with nest thermal properties, creating a system where female behaviour and architectural investment jointly regulate embryonic temperature exposure.

### Implications for Understanding Avian Reproduction

This integrated analysis demonstrates that reproductive success in songbirds emerges from coordinated adjustment across morphological, behavioural, and architectural traits. The findings have important implications for understanding how species may adapt to climate change through adjustments in breeding phenology, nest construction, and incubation behaviour.

### Summary

This analysis examines seasonal variation in reproductive investment through quantification of nest structural properties and egg characteristics across the breeding season. ### Interpretation

**Egg Mass Increases Dramatically Across Breeding Season:** Female mockingbirds invest progressively more in individual egg size, with July eggs approximately 0.89g heavier than March eggs (21% increase). This trend is highly significant (F = 11.7, p < 0.001) and suggests either selection for larger eggs in favorable late-season conditions or adaptive responses to environmental cues predicting hatching success.

**Nest Height Effects on Reproductive Allocation:** Higher nest placement is associated with lighter eggs (coefficient = -0.008g per cm, p = 0.049), potentially reflecting constraints on female condition at elevated nesting sites or adaptive allocation toward smaller clutches in exposed locations. Importantly, this height effect weakens significantly in late season: the interaction term for July × Nest Height (0.025, p = 0.025) indicates that by July, females in higher nests have largely overcome the height-related penalty, possibly reflecting improved foraging conditions or physiological adaptation.

**Nest Material Investment Declines:** Material composition (grass, leaves, twigs) shows significant decline across months (p < 0.05), yet overall nest mass does not decline correspondingly, suggesting that structural integrity is maintained through compensatory increases in other construction components or through more efficient material arrangement. This dissociation between material quantity and nest structural properties is explored mechanistically in Supplementary Material 4 (thermal properties) and Material 2 (functional incubation behaviour).

**Key Finding:** The parallel increase in egg mass and decrease in nest material investment reveals a sophisticated seasonally-adaptive reproductive strategy in which females progressively redirect resources from nest elaboration toward egg provisioning, likely in response to improving environmental conditions and food availability as the breeding season progresses.

**Biparental Coordination:** The extraordinary thermal coupling (R² = 0.754) and near-perfect visit frequency coordination (R² = 0.999) reveal that incubation behaviour is not independent between partners but highly synchronized. This synchronization is mechanistically explained in Supplementary Material 3, which demonstrates that incubation timing follows strong circadian rhythms (χ² = 9,925.4) that likely drive the observed biparental coordination.

**Detection Method:** Automated identification of the pattern Nocturnal→Off→Nocturnal within early morning (01:00-07:00) or night (18:00-24:00) windows.

**Results:**

**Interpretation:** Approximately 40% of nests exhibit brief nocturnal feeding or maintenance interruptions concentrated in early morning hours (1-3 AM), suggesting a circadian-driven maintenance or feeding window. The concentration of these interruptions in early morning rather than throughout the night indicates that females employ a specific adaptive strategy: maintaining nearly uninterrupted incubation throughout most of the night while permitting brief provisioning bouts during a narrow temporal window.

### Interpretation

**Overwhelming Dominance of Circadian Periodicity:** The circadian term (χ² = 9,925.4, edf = 17.3/18, p < 2e-16) is by far the strongest effect in the model, approximately 1.5-fold larger than the seasonal interaction term. The effective degrees of freedom close to the reference dimension (17.3 vs 18) indicates that virtually all available circadian complexity is captured by the smooth function, revealing a sophisticated within-day incubation pattern with multiple peaks and troughs throughout the 24-hour cycle.

**Weak Seasonal Main Effect with Strong Seasonal Modulation:** While the marginal effect of season is weak (χ² = 3.282, p = 0.070), the circadian × season interaction is highly significant (χ² = 6,823.2, p = 2.39e-06). This pattern indicates that absolute incubation propensity does not vary strongly across the breeding season, but rather the timing and intensity of within-day incubation bouts change substantially from early to late breeding season. This finding complements Supplementary Material 2's observation of shorter recesses in late season: females maintain high incubation attendance regardless of season, but the temporal distribution of effort shifts seasonally.

**Low Individual Heterogeneity:** Both nest and nest-day random effects show high p-values (p > 0.89), indicating minimal contribution to explaining variation beyond the fixed and smooth effects. The circadian and seasonal patterns thus represent population-level properties that generalize across individuals, rather than heterogeneous individual preferences. This suggests strong constraint on incubation timing, possibly through physiological demand signals from embryos or through shared environmental (light) cues.

**Early-Morning Nocturnal Interruptions:** The detection of brief nocturnal interruptions in 40% of nests, concentrated at ~1:22 AM, reveals an overlooked aspect of incubation behaviour. These interruptions may reflect a specific adaptive window when females must balance embryonic heating demands against self-maintenance (feeding, preening). The early-morning timing suggests females employ a strategy of maintaining uninterrupted nocturnal incubation until the coldest period (pre-dawn) has passed, then briefly interrupt to restore physiological reserves before resuming continuous incubation as ambient temperature begins rising with approaching dawn.

## Supplementary Material 4: Nest Mass and Thermal Insulation

### Summary

This analysis quantifies how nest mass influences egg cooling rates during incubation recesses through controlled experimental cooling trials. Exponential cooling models were fitted to 157 temperature traces from artificial nests of three mass categories (8g, 22g, 35g), estimating cooling constants and time-to-physiological-zero for each nest type.

|-----------|----------------------|-------------------|----|----|----------|
| 8g (baseline) | 0.001438 | 17.45 min | 0.944 | - | - |
| 22g | 0.001219 | 19.41 min | 1.225 | +1.96 min (+11%) | 0.248 |
| 35g | 0.001195 | 20.99 min | 1.209 | +3.55 min (+20%) | 0.011 |

| Metric | Result | Interpretation |
|--------|--------|----------------|
| Sample size | N = 157 traces | Robust experimental dataset |
| Model Family | Gaussian GLM | Appropriate for continuous temperature data |
| Model R² | 0.054 (Pseudo) | Nest type explains modest variation |
| 35g vs 8g contrast | Significant (p = 0.011) | Only this pairwise difference reaches significance |
| Effect size (35g benefit) | 20% insulation improvement | Biologically meaningful for embryo development |
| Theoretical time-to-26°C (18°C ambient) | 8g: 11.7 min, 22g: 13.8 min, 35g: 14.1 min | Ranking consistent with experimental results |

### Experimental Design and Data Processing

**Methodology:**
- Artificial nests of three discrete mass categories (8g, 22g, 35g) constructed to match natural nest material composition
- Eggs pre-heated to ~40°C (approximate natural incubation temperature)
- Cooling traced at high temporal resolution (~1 second intervals) under ambient laboratory conditions
- Cooling phase isolated from moment when egg temperature was closest to 40°C
- Analysis window: first 2,600 seconds (~43 minutes) to standardize across replicates

**Data Quality Controls:**
- Removed traces showing unrealistic temperatures (>55°C), indicating equipment malfunction
- Retained 157 of 158 valid traces after quality filtering
- Ambient temperature measured throughout each trial and used as covariate in cooling model

### Exponential Cooling Model

**Model Equation:**
$$T(t) = T_{env} + (T_0 - T_{env}) e^{-kt}$$

where:
- T(t) = egg temperature at time t (seconds)
- T_env = ambient (environmental) temperature
- T₀ = initial egg temperature
- k = cooling constant (rate parameter)

**Interpretation of k:**
- Higher k = faster cooling (less insulation)
- Lower k = slower cooling (better insulation)
- k is inversely related to nest insulation quality

### Primary Model Results: Time to Physiological Zero Temperature

**Model Specification:**
```
Time_to_26°C ~ C(Nest_Mass)
Family: Gaussian | Link: Identity
```

**Coefficients:**

### Theoretical Cooling Under Standardized Conditions

To facilitate comparison independent of experimental ambient temperature variation, theoretical cooling times were calculated assuming standardized conditions (initial temperature T₀ = 40°C, ambient temperature T_env = 18°C, target temperature T_target = 26°C):

**Formula:**
$$t_{26} = -\frac{1}{k} \ln\left(\frac{T_{target} - T_{env}}{T_0 - T_{env}}\right) = -\frac{1}{k} \ln\left(\frac{26 - 18}{40 - 18}\right) = -\frac{1}{k} \ln(0.364)$$

**Results:**

| Nest Type | Theoretical Time to 26°C (at 18°C ambient) | % Increase vs 8g |
|-----------|-------------------------------------------|-----------|
| 8g | 11.72 minutes | baseline |
| 22g | 13.83 minutes | +18% |
| 35g | 14.11 minutes | +20% |

This theoretical analysis confirms the ranking observed in experimental data, demonstrating robust insulation benefits of larger nests across varying ambient conditions.

### Interpretation

**Nest Mass as Thermoregulatory Investment:** Heavier nests provide superior insulation against egg cooling during incubation recesses. The 35g nests achieve approximately 20% slower cooling rates compared to 8g nests, a biologically significant difference given typical recess durations of 6-10 minutes (Supplementary 2). This thermoregulatory benefit translates to approximately 3.5 additional minutes before eggs cool to physiological zero temperature, providing a meaningful buffer against developmental interruption.

**Threshold Effect in Nest Mass:** The significant difference between 35g and 8g nests (p = 0.011) coupled with the non-significant 22g intermediate value suggests a threshold effect: modest nest mass increases from 8g to 22g provide marginal insulation benefits, while substantial investment beyond 22g yields disproportionate improvements. This non-linear relationship may explain the seasonal decline in material investment observed in Supplementary 1: females may reduce material investment in late season not because insulation becomes less important, but because optimal nest mass can be achieved through more efficient construction using fewer but more strategically-placed materials.

**Implications for Egg Investment and Breeding Strategy:** The thermoregulatory benefits of heavier nests provide an alternative mechanistic explanation for the observed increase in egg mass across the breeding season (Supplementary 1). If nests become lighter (reduced material) but females compensate through larger eggs, the combined effect of larger, better-provisioned embryos in more efficient nests could represent an adaptive optimization of reproductive allocation as the season progresses and environmental conditions improve.

## Integrated Analysis: How the Four Supplementaries Work Together

### Conceptual Framework

The four supplementary materials reveal three interconnected biological systems that jointly govern reproductive success in Northern Mockingbirds:

### System 1: Nest Architecture and Thermal Properties

**Supp 1 → Supp 4 Linkage:**

Supplementary 1 documents that nest material investment **declines** across the breeding season despite maintained structural integrity. Supplementary 4 mechanistically explains this paradox: even light nests (8g) provide substantial insulation, and diminishing returns in insulation are achieved beyond ~22g nest mass. 

**Adaptive Interpretation:** Females allocate resources away from nest elaboration in late season not because thermal properties become unimportant, but because they achieve sufficient insulation through increasingly efficient nest construction. This resource reallocation permits investment in larger, better-provisioned eggs (Supp 1: +0.89g by July) that may confer developmental advantages offsetting the marginal thermoregulatory gains of heavier nests.

### System 2: Egg Provisioning and Incubation Behavior

**Supp 1 → Supp 2 Linkage:**

Supplementary 1 shows that larger eggs are laid in late season. Supplementary 2 demonstrates that incubation attendance increases correspondingly (87.0% March → 89.9% July), and recess duration shortens (8.3 min → 6.9 min). 

**Mechanistic Explanation:** Larger eggs require more intensive thermal management to maintain optimal developmental temperature. The increased incubation attendance and shortened recesses represent female behavioural accommodation to the increased physiological demands of larger embryos. ## Technical Appendix: Model Types and Diagnostic Approaches

### Generalized Linear Models (GLM)

**Definition:** Extension of linear regression to non-normally distributed response variables through link functions.

**Families used in this analysis:**
- **Gaussian:** For continuous variables with normal error distribution (egg mass, nest mass, cooling times)
- **Poisson:** For count data (clutch size, visit frequency)
- **Gamma:** For positive continuous variables with right-skewed distribution (recess duration)
- **Binomial:** For binary responses (incubation attendance as success/failure)

**Diagnostics:**
- Residual deviance vs df: Should be approximately equal for well-specified models
- Pearson residuals: Standardized residuals that should be roughly normally distributed
- Overdispersion parameter: Scale parameter >1 indicates extra-binomial variation

### Linear Mixed-Effects Models (LMM)

**Definition:** Regression models incorporating both fixed effects and random effects to account for hierarchical data structure.

**Syntax used in Supp 2:**
```
Response ~ Fixed_Predictors + (1|Random_Intercept)
```

**Random Intercepts in this analysis:**
- Nest_ID: Permits each nest to have different baseline incubation attendance
- Pair_ID: Permits each mating pair to have different baseline behaviour
- Day: Accounts for temporal clustering of observations within days

**Diagnostics:**
- Q-Q plots of random effects: Should approximate normality
- Intraclass correlation (ICC): Proportion of variance explained by random effects
- Marginal vs conditional R²: Marginal = fixed effects only, Conditional = including random effects

### Generalized Additive Mixed Models (GAMM)

**Definition:** Flexible regression framework where predictor effects are modeled as smooth functions rather than linear terms.

**Model Formula (Supp 3):**
```
incubating ~ s(minute_of_day, bs="cc", k=20) +
             s(DOY, k=10) +
             ti(minute_of_day, DOY, bs=c("cc","tp"), k=c(20,10)) +
             s(Nest_ID, bs="re") +
             s(NestDay_ID, bs="re")
```

**Basis Function Types:**
- **bs="cc"** (cyclic cubic spline): For circular/periodic variables (time-of-day cycles at 1440 minutes)
- **bs="tp"** (thin-plate spline): Flexible basis for non-periodic smooth effects (day-of-year trend)
- **bs="re"** (random effect smooth): Random intercepts using spline penalty framework

**Basis Dimension (k parameter):**
- Specifies maximum complexity of smooth function
- k=20 for circadian (adequate for capturing ~20 unique features in 24-hour cycle)
- k=10 for seasonal (adequate for capturing ~10 unique features across 5-month season)
- Automatically penalized if not needed (edf < k)

**Key Diagnostics:**
- **Effective Degrees of Freedom (edf):** Actual complexity used; edf << k suggests overspecification
- **k-index:** Diagnostic statistic for basis dimension adequacy (values <1 suggest k too low)
- **Concurvity:** Measures collinearity among smooth terms (values near 0 indicate independence)
- **Convergence:** fREML optimiser should achieve near-zero gradient

**Significance Tests:**
Approximate chi-squared test of each smooth term H0: smooth term = 0 (linear)

### Post-Hoc Comparisons

**Tukey Honest Significant Difference (HSD):**
- Pairwise comparisons of means with family-wise Type I error correction
- Used in Supp 4 to compare cooling times among 3 nest mass categories
- p-values reported are adjusted for multiple testing
- Formula: Mean_Difference ± qα × SE(contrast)

**Circular Statistics (Supp 3):**
- **Circular mean angle:** Mean direction on a circle (for timing of nocturnal interruptions)
- **Rayleigh test:** Tests whether angles are randomly distributed vs concentrated in one direction
- Large test statistic and small p-value indicate non-random clustering of events

## Glossary of Technical Terms

**Adaptive Radiation:** Evolutionary diversification of species in response to environmental opportunities (not used here, but relevant context for seasonal adaptation)

**Binomial Family:** Probability distribution for binary (0/1) response variables; used for presence/absence data

**Biparental Cooperation:** Joint investment by both parents in offspring care; quantified here through coordination of incubation attendance

**Circadian Rhythm:** Behavioral or physiological pattern repeating approximately every 24 hours under entrainment to light-dark cycles

**Clutch Size:** Number of eggs laid in a single breeding attempt; typically 3-4 eggs in mockingbirds

**Concurvity:** In GAMMs, collinearity among smooth terms; values near 0 indicate independence

**Confidence Interval (CI):** Range of values likely to contain the true parameter value with specified probability (95% CI commonly reported)

**Cooling Constant (k):** Rate parameter in exponential cooling model; inverse relationship with insulation quality

**Covariates:** Predictor variables included in statistical models; may include confounders or predictors of interest

**Day-of-Year (DOY):** Julian day number (1-365) indicating seasonal progression; March ≈ 60-90, July ≈ 180-210

**Deviance:** Generalization of sum-of-squares residuals for GLMs; smaller deviance indicates better fit

**Deviance Explained:** Proportion of total deviance explained by the model; analogous to R² for normal regression

**Effective Degrees of Freedom (edf):** In GAMMs, the effective complexity of a smooth term; lower edf indicates greater regularization

**Embryonic Development:** Process of egg maturation and fetal growth; affected by incubation temperature, duration, and oxygen availability

**Exponential Cooling Model:** Mathematical model T(t) = Tenv + (T0-Tenv)exp(-kt) describing temperature decay in thermal systems

**Family-wise Error Rate:** Probability of committing at least one Type I error across multiple statistical tests; corrected by Tukey HSD

**Generalized Additive Mixed Model (GAMM):** Extension of GAM incorporating random effects for hierarchical data

**Generalized Linear Model (GLM):** Regression framework accommodating non-normal response distributions and link functions

**Goodness-of-Fit:** Assessment of model adequacy through comparison of predicted vs observed values

**Hessian Matrix:** Matrix of second partial derivatives; used to assess convergence of optimization algorithms

**Incubation Attendance:** Proportion of time female(s) spend at the nest maintaining egg warmth; typically >85% in oscines

**Incubation Recesses:** Off-nest periods during incubation when females forage, drink, preen, or maintain self-condition

**Interaction Term:** Product of two predictor variables; tests whether effect of one predictor depends on level of another

**Intraclass Correlation (ICC):** Proportion of variance attributable to clustering; values near 0 indicate weak clustering

**k-index:** Diagnostic in GAMMs; values substantially <1 suggest basis dimension (k) too low

**Kolmogorov-Smirnov Test:** Non-parametric test comparing observed distribution to reference distribution

**Link Function:** Mathematical function relating linear predictor to response distribution mean; identity (Gaussian), log (Poisson), logit (Binomial)

**Logit Function:** log(p/(1-p)); transforms probability (0-1) to unbounded linear scale; inverse is logistic function

**Mean Egg Mass:** Average weight of eggs in a clutch; typically 4.2-4.8g in Northern Mockingbirds

**Minute-Level Resolution:** Behavioral observations recorded at 1-minute temporal intervals; permits detection of fine-scale patterns

**Mixed-Effects Model:** Regression with both fixed effects (apply to all observations) and random effects (vary by group)

**Model Residuals:** Differences between observed and predicted values; should be randomly distributed around zero for well-specified models

**Multicolinearity:** High correlation among predictor variables; can inflate standard errors and reduce precision

**Nest Height:** Vertical distance from ground to nest structure; typically 1.5-8 meters in mockingbirds

**Nest Mass:** Total weight of constructed nest; function of material amount and density

**Nest Material:** Composition of nest construction (grasses, leaves, twigs, feathers, human-made materials); affects thermal and structural properties

**Newton's Law of Cooling:** Principle that cooling rate is proportional to temperature difference: dT/dt ∝ (T - Tenv)

**Nocturnal Incubation:** Continuous incubation behaviour during nighttime hours; females often incubate alone at night

**Nonlinear Relationship:** Association between variables that is not accurately described by a straight line; modeled using smooth functions in GAMs

**Optimization Algorithm:** Computational method for finding parameter values that maximize likelihood; fREML, REML, MLE commonly used

**Overdispersion:** Extra-binomial variation in count or binary data (Scale parameter >1); indicates inadequate model specification

**p-value:** Probability of observing result as extreme or more extreme under null hypothesis; threshold α = 0.05 commonly used

**Phenology:** Timing of biological events (breeding, migration, molt); responsive to environmental cues (temperature, photoperiod, food availability)

**Physiological Zero Temperature:** Temperature below which embryonic development ceases; approximately 26°C for avian eggs

**Poisson Family:** Probability distribution for count data; assumes mean = variance

**Predicted Values:** Estimated response values from fitted model; obtained using model coefficients and new predictor values

**Rayleigh Test:** Circular statistics test; null hypothesis is that angles are uniformly distributed

**R²:** Coefficient of determination; proportion of variance in response explained by model predictors; ranges 0-1

**Reference Level (Baseline):** Category of categorical variable coded as zero in contrast coding; all other categories compared to reference

**Regularization:** Penalty applied to model complexity; prevents overfitting and stabilizes parameter estimates

**Recess Duration:** Length of off-nest periods; typically 5-15 minutes in incubating females

**Restricted Maximum Likelihood (REML):** Parameter estimation method accounting for uncertainty in fixed effects; preferred for mixed models

**Seasonal Variation:** Systematic changes in biological traits across breeding season (spring-summer); response to environmental and developmental cues

**Significance Level (α):** Threshold p-value below which null hypothesis is rejected; α = 0.05 (5% Type I error) standard in biology

**Smooth Function:** Nonlinear function estimated from data; in GAMs, represented as linear combination of basis functions

**Spline:** Piecewise-polynomial function that is smooth (continuous, continuous derivatives); used as basis in GAMs

**Standard Error (SE):** Standard deviation of sampling distribution of parameter estimate; inverse relationship with sample size and data variability

**Tensor Product Interaction:** In GAMs, interaction between two smooth terms; ti() function estimates interaction net of main effects

**Thermal Coupling:** Coordination between biparental partners in maintaining embryonic temperature; quantified through correlation of incubation duration

**Thin-Plate Spline (tp):** Flexible smooth basis with good default properties; often used for non-periodic effects

**Time-of-Day:** Position in 24-hour cycle; minute_of_day ranges 0-1439; time-of-day cyclic with period = 1440 minutes

**Visit Frequency:** Number of provisioning or incubation visits per unit time; typically measured as visits per hour or visits per day

**Zenith Angle:** Angle of sun above horizon; primary environmental cue entraining circadian rhythms
