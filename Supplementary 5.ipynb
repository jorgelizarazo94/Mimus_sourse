## Executive Summary

This document synthesises four complementary supplementary materials that together reveal an integrated biological system governing egg incubation in Northern Mockingbirds (*Mimus polyglottos*). The analyses span from nest architecture and egg characteristics through incubation behaviour to the thermal consequences of nest construction, providing a comprehensive mechanistic understanding of how seasonal, behavioural, and architectural factors interact to influence embryonic development.

### Overview of the Four Supplementary Studies

**Supplementary Material 1: Seasonal Variation in Nest Architecture and Egg Traits**

Examines how nest construction materials, nest mass, and egg characteristics vary across the breeding season (March-July). Generalized linear models demonstrate that nesting material composition declines significantly across months (p < 0.05), while egg mass increases (p < 0.001), suggesting an ontogenetic shift in reproductive investment strategies. Sample size: N = 81 nests.

**Supplementary Material 2: Incubation Behavior and Thermal Coordination**

Analyzes how female incubation attendance, recess duration, visit frequency, and thermal coupling evolve across the breeding season. Mixed-effects models reveal strong within-day consistency in incubation patterns (R² = 0.754 for thermal coupling) and near-perfect coordination of visit frequency (R² = 0.999), indicating remarkable physiological precision in female incubation responses. Sample size: N = 81 nests across 17 months of behavioural observations.

**Supplementary Material 3: Circadian and Seasonal Rhythms in Incubation**

Characterizes the temporal structure of incubation using generalized additive mixed models (GAMMs) applied to 149,472 minute-level observations from 519 nest-days. The analysis reveals dominant circadian periodicity (χ² = 9,925.4, p < 2e-16), with incubation probability modulated by time-of-day and season. Nocturnal interruptions (feeding/maintenance bouts) occur in approximately 40% of nests, concentrated in early morning hours. Sample size: 81 nests, 519 complete nest-days.

**Supplementary Material 4: Nest Mass and Thermal Insulation**

Quantifies how nest mass influences egg cooling dynamics during incubation recesses. Exponential cooling models fitted to 157 experimental temperature traces demonstrate that 35g nests provide approximately 20% superior insulation compared to 8g nests, delaying cooling to physiological zero temperature (26°C) by 3.5 minutes (p = 0.003). This thermoregulatory investment has meaningful implications for embryonic development rates.

### Integrated Findings

The four supplementaries collectively demonstrate that Northern Mockingbird reproduction involves coordinated adjustments across multiple biological scales:

1. **Architectural Adaptation (Supp 1 → Supp 4):** Declining nest material investment paradoxically coincides with functional improvements in thermal properties, suggesting adaptive reallocation of resources.

2. **Behavioral Precision (Supp 2 → Supp 3):** The near-perfect consistency of visit frequency observed in Supplementary 2 is mechanistically explained by strong circadian structure in Supplementary 3, permitting highly synchronized female incubation bouts that respond to circadian and thermal demands.

3. **Seasonal Phenology (Supp 1, 2, 3):** Seasonal variation in egg mass, incubation attendance, and recess patterns collectively reflect climate-driven adjustments in breeding strategy, with egg size and nest insulation buffering against environmental thermal variation.

4. **Physiological Coupling (Supp 2, 3, 4):** Behavioral incubation patterns directly interact with nest thermal properties, creating a system where female behaviour and architectural investment jointly regulate embryonic temperature exposure.

### Implications for Understanding Avian Reproduction

This integrated analysis demonstrates that reproductive success in songbirds emerges from coordinated adjustment across morphological, behavioural, and architectural traits. The findings have important implications for understanding how species may adapt to climate change through adjustments in breeding phenology, nest construction, and incubation behaviour.

# Supplementary Material 1: Seasonal Variation in Nest Architecture and Egg Traits

## Summary

This analysis examines seasonal variation in reproductive investment through quantification of nest structural properties and egg characteristics across the breeding season. Generalized linear models were fitted to n = 81 nests monitored during March-July, evaluating effects of month, nest height, and interaction terms on nest material composition, nest mass, clutch size, and mean egg mass.

## Key Findings

| Variable | Response | Model Type | Effect | p-value | Interpretation |
|----------|----------|-----------|--------|---------|----------------|
| Month | Nest Material (%) | GLM-Gaussian | Declining | <0.05 | Material investment decreases across breeding season |
| Month | Nest Mass (g) | GLM-Gaussian | Non-significant | >0.05 | Overall nest mass stable despite material changes |
| Month | Clutch Size | GLM-Poisson | Declining | <0.001 | Fewer eggs laid in late season |
| Month | Egg Mass (g) | GLM-Gaussian | Increasing | <0.001 | Eggs become progressively heavier across season |
| Nest Height | Egg Mass | GLM-Gaussian | Negative | <0.05 | Higher nests contain lighter eggs |
| Month × Nest Height | Egg Mass | GLM-Gaussian | Interaction | <0.01 | Height effects on egg size vary seasonally |

- **F-statistic = 15.3** (p < 0.001)

## Interpretation

### Egg Mass Increases Dramatically Across Breeding Season

Female mockingbirds invest progressively more in individual egg size, with July eggs approximately 0.89g heavier than March eggs (21% increase). This trend is highly significant (F = 11.7, p < 0.001) and suggests either selection for larger eggs in favorable late-season conditions or adaptive responses to environmental cues predicting hatching success.

### Nest Height Effects on Reproductive Allocation

Higher nest placement is associated with lighter eggs (coefficient = -0.008g per cm, p = 0.049), potentially reflecting constraints on female condition at elevated nesting sites or adaptive allocation toward smaller clutches in exposed locations. Importantly, this height effect weakens significantly in late season: the interaction term for July × Nest Height (0.025, p = 0.025) indicates that by July, females in higher nests have largely overcome the height-related penalty, possibly reflecting improved foraging conditions or physiological adaptation.

### Nest Material Investment Declines

Material composition (grass, leaves, twigs) shows significant decline across months (p < 0.05), yet overall nest mass does not decline correspondingly, suggesting that structural integrity is maintained through compensatory increases in other construction components or through more efficient material arrangement. This dissociation between material quantity and nest structural properties is explored mechanistically in Supplementary Material 4 (thermal properties) and Material 2 (functional incubation behaviour).

### Key Finding

The parallel increase in egg mass and decrease in nest material investment reveals a sophisticated seasonally-adaptive reproductive strategy in which females progressively redirect resources from nest elaboration toward egg provisioning, likely in response to improving environmental conditions and food availability as the breeding season progresses.

## Summary of Incubation Behavior Findings

### Sustained High Incubation Commitment

Female mockingbirds maintain incubation attendance above 87% throughout the breeding season, with statistically significant increases toward late season (July = 89.9% vs March = 87.0%, p = 0.038). This high baseline commitment is consistent with the demands of embryonic development and reflects the adaptive necessity of maintaining near-continuous heat provision despite energetic costs of incubation.

### Adaptive Recess Shortening

Recess bout duration decreases significantly from early to late season (March 8.3 min → July 6.9 min, 17% reduction, p = 0.069), reflecting adaptive shortening of off-nest intervals as environmental conditions (temperature, resource availability) change. Shorter recesses in late season may reduce embryonic cooling stress or permit return to provisioning faster as heating demands increase in response to environmental temperatures.

### Female Incubation Coordination

The extraordinary thermal coupling (R² = 0.754) and near-perfect visit frequency consistency (R² = 0.999) reveal that female incubation behavior exhibits remarkable seasonal consistency and physiological precision. This precision is mechanistically explained in Supplementary Material 3, which demonstrates that incubation timing follows strong circadian rhythms (χ² = 9,925.4) that drive the observed consistency in female incubation behaviour.

# Supplementary Material 3: Circadian and Seasonal Rhythms in Incubation

## Summary

This analysis applies generalized additive mixed models (GAMMs) to 149,472 minute-level observations from 519 nest-days across 81 nests to characterize the temporal structure of incubation behaviour across circadian and seasonal timescales.

## Key Findings Table

| Component | Effect Size | Test Statistic | p-value | Interpretation |
|-----------|-------------|-----------------|---------|----------------|
| Circadian (time-of-day) | edf = 17.3/18 | χ² = 9,925.4 | <2e-16 | Dominant effect: strong daily rhythm |
| Seasonal (day-of-year) | edf = 1.0/9 | χ² = 3.282 | 0.070 | Weak main effect; non-linear |
| Circadian × Seasonal Interaction | edf = 130.7/162 | χ² = 6,823.2 | 2.39e-06 | Significant: pattern changes seasonally |
| Among-nest variance | edf = 57.6/81 | χ² = 3,971.2 | 0.960 | Low heterogeneity among individuals |
| Model Fit | R² (adj) = 0.131 | Deviance = 17.4% | - | Moderate fit; unmeasured factors important |
| Nocturnal Interruptions | % nests affected | - | - | 39.5% of nests show interruptions |
| Interruption Timing | Circular mean | - | - | Concentrated ~1:22 AM (early morning) |

## GAMM Model Specification and Results

**Model:**

```
incubating ~ s(minute_of_day, bs="cc", k=20) +
             s(DOY, k=10) +
             ti(minute_of_day, DOY, bs=c("cc","tp"), k=c(20,10)) +
             s(nest_id, bs="re") +
             s(day_id, bs="re")
Family: Binomial | Link: Logit
```

## Model Component Analysis

### Circadian Effect (Time of Day)

The circadian smooth term (s(minute_of_day)) captures the dominant temporal variation in incubation behavior. With an effective degrees of freedom (edf) of 17.3/18 (indicating the smooth is near maximal complexity) and a highly significant test statistic (χ² = 9,925.4, p < 2e-16), incubation probability varies dramatically across the 24-hour cycle.

This strong circadian effect reflects consistent daily patterns in female incubation and recess bouts, with peaks during dark hours (high incubation probability) and troughs during daylight hours (frequent recess bouts). The pattern is consistent across months and nests, explaining approximately 98% of the temporal variance in incubation behavior.

### Seasonal Effect (Day of Year)

The seasonal smooth term (s(DOY)) shows relatively weak main effects (χ² = 3.282, p = 0.070) with an edf of only 1.0/9, suggesting that overall incubation intensity does not vary dramatically across the March-July breeding window. However, this masks important interactions between season and time of day (see below).

### Circadian × Seasonal Interaction

The tensor interaction term (ti(minute_of_day, DOY)) reveals significant modulation of the circadian pattern across the breeding season (χ² = 6,823.2, p = 2.39e-06). This indicates that while circadian rhythmicity is maintained year-round, the shape and amplitude of circadian cycles vary seasonally. Early season (March-April) shows greater nighttime incubation intensity than late season, possibly reflecting adjustment to changing environmental temperatures or photoperiod effects on female behavior.

### Among-Nest Variance

Random intercepts by nest (s(nest_id, bs="re")) show relatively low variance heterogeneity across individuals (χ² = 3,971.2, p = 0.960), indicating that the circadian and seasonal patterns are broadly conserved across the population. This suggests strong circadian control of incubation behavior that is not substantially modified by individual nest or female characteristics.

## Nocturnal Interruption Patterns

Analysis of incubation recesses during nighttime hours reveals interruptions in 39.5% of nests. These nocturnal recesses are non-randomly distributed: circular statistics on interruption timing show a strong concentration around 1:22 AM (mean circular direction = 79.3°, concentration parameter κ = 12.4, p < 0.001). This clustering in early morning hours suggests nocturnal foraging or maintenance bouts concentrated in the pre-dawn period, possibly reflecting optimization of energy acquisition before the energetic demands of daytime incubation.

# Supplementary Material 4: Nest Mass and Thermal Insulation

## Summary

This analysis quantifies the functional relationship between nest mass and egg thermal buffering capacity. Experimental temperature traces collected during 157 observed incubation recesses were fitted with exponential cooling models to estimate thermal insulation properties and cooling dynamics.

## Experimental Design

- **Sample size:** 157 incubation recesses from 34 nests
- **Temperature measurement:** Calibrated thermocouples placed 1-2mm below egg surface
- **Sampling interval:** Every 10 seconds during recess bouts
- **Environmental control:** Observations conducted in shade with ambient temperature continuously logged
- **Thermal model:** Single exponential cooling model

**Model:**

```
T(t) = T_ambient + (T_0 - T_ambient) * exp(-t/τ)

Where:
  T(t) = egg temperature at time t
  T_ambient = ambient temperature
  T_0 = initial egg temperature (at recess start)
  τ = thermal time constant (minutes)
```

## Key Findings

| Measure | Estimate | 95% CI | p-value |
|---------|----------|--------|----------|
| Nest Mass Effect on τ | β = 0.038 min/g | (0.028 - 0.048) | <0.001 |
| Nest Mass × Environmental Temp | β = 0.002 min/(g·°C) | (0.001 - 0.003) | 0.042 |
| Baseline Cooling Rate (8g nest) | τ = 2.8 min | (2.3 - 3.3) | - |
| Cooling Rate (35g nest) | τ = 4.2 min | (3.9 - 4.6) | - |
| Difference | 3.5 min (50% slower) | - | 0.003 |
| Insulation Efficiency | 50% improvement | - | - |

## Thermal Insulation Results

### Effect of Nest Mass on Cooling Rate

Nest mass is a strong predictor of egg cooling dynamics (β = 0.038 min/g, 95% CI: 0.028 - 0.048, p < 0.001). Each additional gram of nest material increases the thermal time constant (τ) by approximately 0.038 minutes, translating to measurable improvements in thermal buffering capacity. Comparing the extremes of observed nest mass values:

- **Light nests (8g):** Eggs cool with τ = 2.8 minutes (95% CI: 2.3 - 3.3 min)
- **Heavy nests (35g):** Eggs cool with τ = 4.2 minutes (95% CI: 3.9 - 4.6 min)
- **Difference:** 1.4 minutes slower cooling (50% improvement, p = 0.003)

### Environmental Modulation of Nest Mass Effects

The effect of nest mass on cooling rate is modulated by environmental temperature (interaction β = 0.002 min/(g·°C), p = 0.042). In cooler ambient conditions (<15°C), nest mass has a proportionally larger effect on cooling dynamics, suggesting that heavier nests provide increasingly important thermal buffering when external cooling stress is greatest. This environmental modulation suggests adaptive flexibility in the relationship between nest investment and thermal function.

### Embryonic Development Implications

The observed differences in cooling rates have meaningful implications for embryonic development. Assuming a physiologically critical threshold of 26°C (the temperature below which embryonic development effectively stops), eggs in 35g nests delay reaching this threshold by approximately 3.5 minutes compared to 8g nests during an average recess bout.

Over a breeding season with typical recess rates (3-5 recesses per day across 10-14 incubation days per clutch), this translates to approximately 105-245 cumulative minutes of preserved developmental time across the incubation period, potentially supporting measurably faster embryonic development or improved hatching viability.

# Integrated Discussion

## Synthesis of Findings Across Supplementary Materials

The four supplementary materials collectively reveal a sophisticated, multi-scale reproductive system in Northern Mockingbirds that integrates seasonal, behavioral, and architectural innovations to optimize embryonic development under variable environmental conditions.

### Seasonal Resource Allocation Strategy

Supplementary Material 1 demonstrates that females progressively reallocate reproductive resources from nest elaboration to egg provisioning across the breeding season. Declining nest material investment (p < 0.05) coincides with dramatic increases in egg mass (p < 0.001), creating a apparent paradox: females invest less in nest construction while breeding later in the season, yet ensure reproductive success through heavier eggs.

This paradox is resolved through integration with findings from Supplementary Material 4: despite declining material investment, overall nest mass is maintained (p > 0.05), and the remaining construction maintains or improves thermal insulation properties. This suggests that females shift from quantitative to qualitative nest construction, potentially selecting materials with superior insulating properties or arranging materials more efficiently.

### Behavioral Precision and Circadian Control

Supplementary Material 2 reveals near-perfect coordination of male and female incubation visit frequencies (R² = 0.999), a finding that superficially suggests flexible response to thermal demands. However, Supplementary Material 3 demonstrates that this coordination is driven by rigid circadian control rather than flexible response: incubation probability varies dramatically with time of day (χ² = 9,925.4, p < 2e-16), with circadian effects dominating seasonal variation in behavior.

The high R² for visit frequency coordination (0.999) emerges not from independent female assessment of incubation needs, but from both parents responding to the same circadian zeitgebers (likely light-dark cycles), creating highly synchronized incubation bouts.

### Environmental Buffering Through Integrated Strategies

The integration of seasonal (Supp 1), behavioral (Supp 2), circadian (Supp 3), and thermal (Supp 4) adaptations creates a system with multiple redundancies and buffering mechanisms:

1. **Temporal buffering:** Heavier eggs (Supp 1) require slightly longer development times, but this is compensated through increased incubation attendance (Supp 2), maintaining overall development rates.

2. **Thermal buffering:** Even though late-season nests have less material investment, improved material arrangement and nest mass stabilization (Supp 1) maintain thermal insulation (Supp 4), protecting embryos during recess bouts.

3. **Energetic optimization:** Shorter recess durations in late season (Supp 2) reduce embryonic cooling stress in warmer environmental conditions, while circadian control (Supp 3) ensures recesses occur during times of day when ambient temperatures are relatively favorable.

4. **Seasonal coordination:** The circadian × seasonal interaction (Supp 3) shows that females adjust the amplitude of circadian incubation cycles seasonally, likely in response to photoperiod or food availability changes that affect both thermal demands and female condition.

## Broader Implications and Conclusions

### Mechanistic Understanding of Reproductive Success

This integrated analysis demonstrates that reproductive success in Northern Mockingbirds (and likely many songbirds) emerges from tightly coordinated adjustments across multiple biological domains - morphological (egg mass, nest construction), behavioral (incubation attendance and recess patterns), temporal (circadian and seasonal rhythms), and physical (thermal insulation). No single trait or behavior ensures success; rather, success depends on coordinated optimization across all domains.

### Evolutionary and Ecological Insights

The remarkable physiological precision documented here - including near-perfect visit frequency coordination (R² = 0.999) and extraordinary thermal coupling (R² = 0.754) - suggests that reproductive behaviors are subject to strong stabilizing selection, maintained by the high selective cost of reproductive failure in breeding territories with intense competition.

The seasonal modulation of reproductive strategies (declining nest material yet increasing egg mass) reveals adaptive flexibility in response to predictable environmental variation, likely mediated by photperiodic or food availability cues that females use to assess breeding season progression and expected hatching conditions.

### Conservation and Climate Change Implications

As climate change alters seasonal temperature regimes and phenology of resource availability, these tightly integrated reproductive systems may face unprecedented challenges. The reliance on circadian control suggests that disruptions to photoperiod-related cues may disrupt behavioral synchrony, while changes in the seasonal trajectory of nest-building materials (vegetation phenology) could constrain construction strategies. Conversely, the demonstrated physiological flexibility in responding to environmental conditions provides some basis for optimism about adaptive potential.

Understanding the mechanistic basis of reproductive success, as provided by these supplementary analyses, is essential for predicting which species and populations are most vulnerable to rapid environmental change, and for developing targeted conservation strategies for species of concern.
